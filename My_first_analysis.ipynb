```python
# Install essential libraries (usually pre-installed, but good practice)
!pip install pandas numpy matplotlib seaborn

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Libraries loaded successfully!")
```

```python
# Download a sample dataset (e.g., tennis match data or sales data)
!wget https://raw.githubusercontent.com/plotly/datasets/master/tips.csv -O tips.csv
```

```python
# Load the dataset (replace 'tips.csv' with your filename)
df = pd.read_csv('tips.csv')

print("✅ Data loaded!")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['total_bill'], kde=True)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='total_bill', y='tip', data=df, hue='sex')
plt.title('Total Bill vs Tip by Gender')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Example: Average tip by gender
df.groupby('sex')['tip'].mean()

# Example: Total bill by day
df.groupby('day')['total_bill'].sum().sort_values(ascending=False)

# Example: Correlation between bill and tip
df['total_bill'].corr(df['tip'])
```
