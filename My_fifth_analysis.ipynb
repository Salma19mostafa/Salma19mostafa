```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
```

```python
tips.head()
tips.columns
len(tips)
tips.isnull().sum()
```

```python
tips.groupby("day")["tip"].mean()
tips.groupby("sex")["tip"].mean()
tips.groupby("time")["tip"].mean()

# Combined comparison
tips.groupby(["day","sex","time"])["tip"].mean()
```

```python
plt.hist(tips["total_bill"], bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(x="day", y="tip", data=tips)
plt.title("Tips by Day")
plt.show()
```

```python
tips.groupby("day")["tip"].mean().idxmax()
```

The average tip varies across days, with one day standing out as the most generous. Male and female customers show slight differences in tipping behavior, while dinner generally yields higher tips than lunch. Overall, the distribution of tips reflects both social patterns and dining context, with the highest average tips concentrated on a specific day of the week.
