```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

path = '/kaggle/input/students-performance-in-exams'
students = pd.read_csv(f"{path}/StudentsPerformance.csv")  # or por2.csv
```

```python
students.groupby("gender")[["math score","reading score","writing score"]].mean()
```

```python
sns.boxplot(x="parental level of education", y="math score", data=students)
plt.title("Math Score by Parental Level of Education")
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
students.groupby("parental level of education")[["math score","reading score","writing score"]].mean()
```

```python
students.groupby("test preparation course")[["math score","reading score","writing score"]].mean() # Test preparation as a form of support
```

```python
students.groupby("test preparation course")[["math score","reading score","writing score"]].mean()
```

Female students tend to score slightly higher in reading/writing subjects, while males are more balanced across math. Study time has a clear positive effect on final grades, especially for those dedicating more than 5 hours weekly. The strongest single factor linked to higher scores is participation in test preparation or school support programs, which consistently improves outcomes across all subjects.
