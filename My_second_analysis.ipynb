```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("Rows where total_bill > 20:")
display(df[df["total_bill"] > 20].head())

print("\nAverage tip by day:")
display(df.groupby("day")["tip"].mean().sort_values(ascending=False))

print("\nAverage total bill by time:")
display(df.groupby("time")["total_bill"].mean())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

plt.figure(figsize=(8,5))
sns.histplot(df["total_bill"], kde=True)
plt.title("Distribution of Total Bill")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
plt.title("Total Bill vs Tip")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="day", y="tip")
plt.title("Tip by Day")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nAverage tip by sex:")
display(df.groupby("sex")["tip"].mean())

print("\nAverage tip by day:")
display(df.groupby("day")["tip"].mean())

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="day", y="tip", errorbar=None)
plt.title("Average Tip by Day")
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="total_bill", y="tip", hue="time")
plt.title("Bill vs Tip by Time")
plt.show()

print("\nFindings:")
print("1. The dataset contains bill, tip, and customer information.")
print("2. Tips vary by day and customer group.")
print("3. Higher bills generally have higher tips.")
```
