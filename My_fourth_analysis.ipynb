```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")
```

```python
iris.groupby("species").mean()
```

```python
sns.scatterplot(x="sepal_length", y="sepal_width", hue="species", data=iris)
plt.title("Sepal Measurements by Species")
plt.show()
```

```python
sns.scatterplot(x="petal_length", y="petal_width", hue="species", data=iris)
plt.title("Petal Measurements by Species")
plt.show()
```

```python
# Drop the categorical column before correlation
numeric_iris = iris.drop(columns=["species"])
sns.heatmap(numeric_iris.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Iris Features")
plt.show()
```

Petal length and width are the strongest features separating the species, with setosa clearly distinct from versicolor and virginica. Sepal measurements show more overlap, making them less reliable for classification. Overall, petal dimensions provide the clearest boundaries between species groups.
