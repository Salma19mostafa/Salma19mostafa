```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
```

```python
titanic.head()
```

```python
titanic.isnull().sum()
```

```python
titanic.groupby("sex")["survived"].mean()
titanic.groupby("class")["survived"].mean()
```

```python
titanic.groupby(["sex","class"])["survived"].mean()
```

```python
sns.barplot(x="sex", y="survived", hue="class", data=titanic)
plt.title("Survival Rates by Sex and Class")
plt.show()
```

Women had a significantly higher survival rate than men, highlighting the “women and children first” policy. First‑class passengers were more likely to survive compared to those in second and third class, showing the impact of social status on safety. Overall, survival was strongly shaped by both gender and class, with third‑class men facing the lowest chances of survival.
