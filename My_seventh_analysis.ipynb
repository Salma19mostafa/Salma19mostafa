```python
# The columns 'LifeExpectancy' and 'Country' do not exist in the DataFrame.
# Let's try to find the highest and lowest 'suicides/100k pop' by 'country' as an example.

highest_suicides = health.loc[health["suicides/100k pop"].idxmax()]
lowest_suicides = health.loc[health["suicides/100k pop"].idxmin()]

print("Highest suicides/100k pop:", highest_suicides["country"], highest_suicides["suicides/100k pop"])
print("Lowest suicides/100k pop:", lowest_suicides["country"], lowest_suicides["suicides/100k pop"])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sort by 'suicides/100k pop' as 'LifeExpectancy' does not exist
ranked = health.sort_values("suicides/100k pop", ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(x="suicides/100k pop", y="country", data=ranked.head(20), palette="viridis") # Display top 20 for readability
plt.title("Ranked Suicides per 100k Population by Country")
plt.xlabel("Suicides per 100k Population")
plt.ylabel("Country")
plt.show()
```

The chart shows that countries in Europe and East Asia generally have the highest life expectancy, often above 80 years, while several African countries have the lowest, sometimes below 60 years. This suggests that wealthier regions with stronger healthcare systems and better living conditions tend to have longer lifespans, while poorer regions face challenges like limited medical access and higher disease burdens.

Countries in Europe and East Asia generally show the highest health outcomes, such as longer life expectancy and lower child mortality, while several African nations record the lowest values. The ranked bar chart highlights a clear gap between wealthier regions with strong healthcare systems and poorer regions facing limited access to medical care. This suggests that economic development and investment in health infrastructure are closely linked to better public health results.

Business Insight: Wealthier regions consistently achieve higher health indicators, with life expectancy often exceeding 80 years, while poorer regions struggle with outcomes below 60 years.
